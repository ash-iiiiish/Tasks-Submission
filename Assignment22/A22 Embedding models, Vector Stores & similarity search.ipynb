# Task4

```python
import numpy as np
from langchain_core.documents import Document
from langchain_community.embeddings import HuggingFaceEmbeddings
from sklearn.metrics.pairwise import cosine_similarity

# Sample chunks
documents = [
    Document(page_content="LangChain enables LLM-powered applications."),
    Document(page_content="Embeddings represent text as vectors."),
    Document(page_content="Vector databases support similarity search."),
]

texts = [doc.page_content for doc in documents]

# Generate embeddings
embeddings=HuggingFaceEmbeddings(model_name='sentence-transformers/all-MiniLM-L6-v2')
doc_vectors = embeddings.embed_documents(texts)

# Search function
def search(query, top_k=3):
    query_vector = embeddings.embed_query(query)
    similarities = cosine_similarity([query_vector], doc_vectors)[0]
    top_indices =similarities.argsort()[-top_k:][::-1]
    return [documents[i] for i in top_indices]

# Test queries
queries = [
    "What are embeddings?",
    "How does similarity search work?",
    "What is LangChain?"
]

for q in queries:
    print(f"\nQuery: {q}")
    for d in search(q):
        print("-", d.page_content)
```

# Task10

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings, OllamaEmbeddings
from langchain_community.vectorstores import FAISS, Chroma

def build_pipeline(
    docs,
    embedding_backend="huggingface",
    vectorstore_backend="faiss"
):
    # -------- Split Documents --------
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=300,
        chunk_overlap=50
    )
    chunks = splitter.split_documents(docs)

    # -------- Embeddings --------
    if embedding_backend == "huggingface":
        emb = HuggingFaceEmbeddings(
            model_name="sentence-transformers/all-MiniLM-L6-v2"
        )
    elif embedding_backend == "ollama":
        emb = OllamaEmbeddings(model="nomic-embed-text")
    else:
        raise ValueError("Unsupported embedding backend")

    # -------- Vector Store --------
    if vectorstore_backend == "faiss":
        store = FAISS.from_documents(chunks, emb)
    elif vectorstore_backend == "chroma":
        store = Chroma.from_documents(chunks, emb)
    else:
        raise ValueError("Unsupported vector store")

    return store
```
