# Task3

```python
def generate_code(prompt):
    return ollama.chat(
        model="codellama:7b",
        messages=[{"role": "user", "content": prompt}]
    )["message"]["content"]


def explain_code(code):
    prompt = f"Explain this code:\n{code}"
    return generate_code(prompt)


def debug_code(code):
    prompt = f"Find and fix bugs in this code:\n{code}"
    return generate_code(prompt)


def optimize_code(code):
    prompt = f"Optimize this code and explain improvements:\n{code}"
    return generate_code(prompt)
```

```python
PROMPTS = {
    "generate": "Write clean, well-documented code for:\n{input}",
    "explain": "Explain the following code step by step:\n{input}",
    "debug": "Identify bugs and fix the following code:\n{input}",
    "optimize": "Optimize this code for performance and readability:\n{input}"
}

def build_prompt(task, text):
    return PROMPTS[task].format(input=text)
```
