# Task4

```python
# Bivariate Histogram
sns.histplot(
    data=df,
    x="G1",
    y="G3"
)
```

```python
# Bivariate KDE
sns.kdeplot(
    data=df,
    x="G1",
    y="G3",
    fill=True
)
```

# Task5

```python
# Pair plot
sns.pairplot(
    df,
    vars=["G1", "G2", "G3", "studytime"],
    hue="sex"
)


```

```python
# Heatmap of correlation
corr = df[["G1", "G2", "G3", "studytime", "absences"]].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")
```

# Task6

```python
# Bar plot
sns.barplot(
    data=df,
    x="sex",
    y="G3"
)
plt.show()


# Box plot
sns.boxplot(
    data=df,
    x="sex",
    y="G3"
)

plt.show()

# Violin plot
sns.violinplot(
    data=df,
    x="sex",
    y="G3"
)

plt.show()

# Count plot
sns.countplot(
    data=df,
    x="sex"
)

plt.show()
```

# Task7

```python
# Regression plot
sns.regplot(
    data=df,
    x="studytime",
    y="G3"
)

plt.show()

# lmplot with hue
sns.lmplot(
    data=df,
    x="studytime",
    y="G3",
    hue="sex"
)

plt.show()
```

# Task8

```python
# FacetGrid
sns.FacetGrid(
    df,
    col="sex"
).map(sns.scatterplot, "studytime", "G3")

plt.show()

# Multi-plot dashboard
sns.relplot(
    data=df,
    x="studytime",
    y="G3",
    hue="sex"
)
sns.catplot(
    data=df,
    x="sex",
    y="G3",
    kind="box"
)
sns.displot(
    df,
    x="G3",
    hue="sex",
    kde=True
)
plt.show()
```
