# Task6

- 1.Large documents exceed an LLM’s context window, making them impossible to process in a single request. Even when they fit, they cause high cost, slower responses, and diluted attention, where important information gets lost among irrelevant text. This leads to lower accuracy, truncation issues, and hallucinations, and also breaks downstream systems like embeddings and retrieval pipelines.
- 2.Chunking splits large documents into smaller, meaningful pieces, enabling efficient retrieval (RAG) and allowing LLMs to see only relevant context. It improves answer accuracy, reduces hallucinations, lowers cost and latency, supports scalability, and enables source attribution and explainability. Chunking is essential for building reliable, production-grade GenAI systems.

```python
from langchain_core.documents import Document


text = """
LangChain is a framework for developing applications powered by language models.

It enables applications that are:
- Data-aware
- Agentic

LangChain provides components such as:
1. Models
2. Prompts
3. Chains
4. Indexes
"""

documents = [Document(page_content=text)]
```

# Task7

```python
from langchain_text_splitters import (
    CharacterTextSplitter,
    RecursiveCharacterTextSplitter,
    MarkdownHeaderTextSplitter
)

# Configure splitter
text_splitter = CharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20
)

# Apply splitter
chunks = text_splitter.split_documents(documents)

# Print results
print("Number of chunks:", len(chunks))
print("\nSample chunk:\n", chunks[0].page_content)
```

```python
from langchain_text_splitters import (
    CharacterTextSplitter,
    RecursiveCharacterTextSplitter,
    MarkdownHeaderTextSplitter
)

markdown_text = """
# Introduction
LangChain is used for LLM applications.

## Features
- Chains
- Agents
- Memory

## Use Cases
Chatbots
"""

headers_to_split_on = [
    ("#", "Header 1"),
    ("##", "Header 2"),
]

markdown_splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=headers_to_split_on
)

md_chunks = markdown_splitter.split_text(markdown_text)

for chunk in md_chunks:
    print(chunk.metadata)
    print(chunk.page_content)
    print("-" * 40)
```

```python
import os
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import (
    DirectoryLoader,
    TextLoader,
    WebBaseLoader
)

def load_and_split_documents(path_or_url):
    """
    Loads documents from a local directory or a web URL,
    splits them into chunks, and returns the chunks.
    """

    # -------- LOAD DOCUMENTS --------
    if path_or_url.startswith("http"):
        loader = WebBaseLoader(path_or_url)
        documents = loader.load()

    elif os.path.isdir(path_or_url):
        loader = DirectoryLoader(
            path_or_url,
            glob="**/*.txt",
            loader_cls=TextLoader
        )
        documents = loader.load()

    else:
        raise ValueError("Input must be a valid directory path or URL")

    # -------- SPLIT DOCUMENTS --------
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=500,
        chunk_overlap=100,
        separators=["\n\n", "\n", " ", ""]
    )

    chunks = splitter.split_documents(documents)
    return chunks
```

```python
local_chunks = load_and_split_documents("./data")
print("Local chunks:", len(local_chunks))
print(local_chunks[0].page_content)
```

 Large PDFs

RecursiveCharacterTextSplitter

 Web data

RecursiveCharacterTextSplitter

3️ Why is chunk overlap important?

Chunk overlap ensures context continuity between chunks.
It prevents loss of meaning at chunk boundaries and improves retrieval accuracy in RAG systems by preserving related information across chunks.
