```python
import pandas as pd 

df = pd.read_csv('C:/Users/kumar/OneDrive/Desktop/TRY-2/Tasks-Submission/Assignment18/7817_1.csv')
df.shape
df.columns
```

```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
text_col = "reviews.text"

# Drop missing values
texts = df[text_col].dropna()
```

# Task1

```python
sentences = texts.iloc[:5].tolist()
vocab = sorted(set(" ".join(sentences).lower().split()))
print("Vocabulary:", vocab)
one_hot_vectors = []

for sentence in sentences:
    words = sentence.lower().split()
    vector = [1 if word in words else 0 for word in vocab]
    one_hot_vectors.append(vector)
one_hot_df = pd.DataFrame(one_hot_vectors, columns=vocab)
one_hot_df
```

# Task2

```python
vectorizer_oh = CountVectorizer(binary=True)
X_oh = vectorizer_oh.fit_transform(sentences)
# Vocabulary
print(vectorizer_oh.vocabulary_)
# Encoded matrix
pd.DataFrame(X_oh.toarray(), columns=vectorizer_oh.get_feature_names_out())
```

# Part2

# Task3

```python
bow_vectorizer = CountVectorizer()
X_bow = bow_vectorizer.fit_transform(texts)
# Vocabulary size
print("Vocabulary size:", len(bow_vectorizer.vocabulary_))
# Sample feature vectors
pd.DataFrame(X_bow[:5].toarray(),
             columns=bow_vectorizer.get_feature_names_out()).head()
```

# Task4

```python
word_freq = np.array(X_bow.sum(axis=0)).flatten()
freq_df = pd.DataFrame({
    "word": bow_vectorizer.get_feature_names_out(),
    "frequency": word_freq
}).sort_values(by="frequency", ascending=False)
# Top 10
freq_df.head(10)
# Least frequent
freq_df.tail(10)
```

Explanation (for report)

Bag of Words captures how many times each word appears in the corpus, ignoring grammar and word order, but preserving frequency information.

# Task5

```python
# Unigrams
uni = CountVectorizer(ngram_range=(1,1))
X_uni = uni.fit_transform(texts)

# Bigrams
bi = CountVectorizer(ngram_range=(2,2))
X_bi = bi.fit_transform(texts)

# Trigrams
tri = CountVectorizer(ngram_range=(3,3))
X_tri = tri.fit_transform(texts)
print("Unigram vocab size:", len(uni.vocabulary_))
print("Bigram vocab size:", len(bi.vocabulary_))
print("Trigram vocab size:", len(tri.vocabulary_))
# Sample vectors
X_bi[:3].toarray()
```
