```python
import pandas as pd
import re
import nltk
import string

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
# Load dataset (example)
df = pd.read_csv('C:/Users/kumar/OneDrive/Desktop/TRY-2/Tasks-Submission/Assignment17/7817_1.csv')

# Select text column
text_col = "reviews.text"

# First 5 text samples
print(df[text_col].head())

# Length of each text
df["text_length"] = df[text_col].astype(str).apply(len)
print(df["text_length"].head())
```

# Task2

```python
def basic_clean(text):
    text = str(text).lower()
    text = re.sub(r'\d+', '', text)                 # remove numbers
    text = text.translate(str.maketrans('', '', string.punctuation))
    text = re.sub(r'\s+', ' ', text).strip()        # remove extra spaces
    return text

df["clean_text_basic"] = df[text_col].apply(basic_clean)

# Compare original vs cleaned
df[[text_col, "clean_text_basic"]].head()
```

# Part2

# Task3

```python
def advanced_clean(text):
    text = str(text).lower()
    text = re.sub(r'http\S+|www\S+', '', text)       # URLs
    text = re.sub(r'\S+@\S+', '', text)              # emails
    text = re.sub(r'<.*?>', '', text)                # HTML tags
    text = re.sub(r'[^a-z\s]', '', text)             # special chars & emojis
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df["clean_text_advanced"] = df["clean_text_basic"].apply(advanced_clean)
```

# Task4

```python
import nltk
nltk.download('punkt_tab')
```

```python
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    words = word_tokenize(text)
    return " ".join([w for w in words if w not in stop_words])

df["text_no_stopwords"] = df["clean_text_advanced"].apply(remove_stopwords)

```

# Task5

```python
slang_dict = {
    "u": "you",
    "gr8": "great"
}

def normalize_text(text):
    # normalize repeated characters
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    
    # replace slang
    words = text.split()
    words = [slang_dict.get(w, w) for w in words]
    return " ".join(words)

df["normalized_text"] = df["text_no_stopwords"].apply(normalize_text)
```

# Part3

# Task6

```python
# Word Tokenization
df["word_tokens"] = df["normalized_text"].apply(word_tokenize)

# Sentence Tokenization (show 3 samples)
for i in range(3):
    print(sent_tokenize(df[text_col].iloc[i]))
```

# Task7

```python
stemmer = PorterStemmer()

def stem_words(tokens):
    return [stemmer.stem(word) for word in tokens]

df["stemmed_words"] = df["word_tokens"].apply(stem_words)

# Compare
df[["word_tokens", "stemmed_words"]].head()
```

# Task8

```python
lemmatizer = WordNetLemmatizer()

def lemmatize_words(tokens):
    return [lemmatizer.lemmatize(word) for word in tokens]

df["lemmatized_words"] = df["word_tokens"].apply(lemmatize_words)

# Compare stemming vs lemmatization
df[["stemmed_words", "lemmatized_words"]].head()
```

# Task9

```python
def nlp_preprocess(text):
    text = advanced_clean(text)
    tokens = word_tokenize(text)
    tokens = [w for w in tokens if w not in stop_words]
    tokens = [lemmatizer.lemmatize(w) for w in tokens]
    return tokens

df["final_clean_text"] = df[text_col].apply(nlp_preprocess)
```

# Task10

- Basic vs Advanced Cleaning

Basic cleaning removes casing, punctuation, numbers

Advanced cleaning removes URLs, emails, HTML, emojis → cleaner NLP input

- Lemmatization vs Stemming

Stemming cuts words aggressively (better → bet)

Lemmatization keeps meaningful words (better → good)

Lemmatization is preferred for NLP models

- Importance of Preprocessing

Reduces noise

Improves model accuracy

Ensures consistent vocabulary

Faster and better feature extraction
